### Arithmetic operators

Integer(int)

```python
a = 20
b = 5
print(a + b)
print(a - b)
print(a * b)
print(a / b)
print(a // b)
print(a % b)
print(a ** 2)
```

Float(float)

```python
a = 10.5
b = 2.5
print(a + b)
print(a - b)
print(a * b)
print(a / b)
print(a // b)
print(a % b)
print(a ** 2)
```

Integer + Float

```python
a = 10
b = 2.5
print(a + b)
print(a - b)
print(a * b)
print(a / b)
print(a // b)
print(a % b)
print(a ** 2)
```

Boolean(bool)

```python
a = True
b = False
print(a + b)
print(a - b)
print(a * b)
print(a / 1)
```

String

```python
line = "-"
print(line * 10)
```
