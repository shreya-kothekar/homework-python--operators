### Comparison Operators

```python
shreya = 22
rutuja = 23
print(shreya > rutuja)
print(shreya < rutuja)
print(shreya == rutuja)
print(shreya != rutuja)
print(shreya >= rutuja)
print(shreya <= rutuja)
```

```python
a = 10
b = 10.0
print(a == b)
print(a != b)
print(a >= b)
print(a <= b)
```

```python
name1 = "shreya"
name2 = "shreya"
print(name1 == name2)
print(name1 != name2)
```
